```python
import numpy as np
```

array函数，两种构建方法

```python
example1=np.zeros((2,3))
print(example1)
example2=np.ones((2,3))
print(example2)
```

输出ndarray对象的四个属性：大小，形状，数据类型，维度

```python
example3=np.array([[1,2,4],[2,5,8]])
print(example3.size)
print(example3.shape)
print(example3.dtype)
print(example3.ndim)
```

改变形状,reshape()函数

```python
print(np.reshape(example3,(-1)))
```

ndarray的算数操作

```python
example4=np.array([[11,22,3,4],[2,33,4,35]])
example5=np.array([[31,4,52,6],[32,9,73,6]])
print(example4<example5)
print(example4+example5)
print(example4*example5)
print(example4/example5)
print(example4**example5)  
print(example4.max())
print(example4.min())
print(example4.sum())
print(example4.max(axis=0))
print(example4.argmax(axis=0))
print(example4.max(axis=1))
print(example4.argmax(axis=1))
```

随机数生成

```python
print(np.random.random_sample(size=[1,2]))
print(np.random.choice(a=[1,2,3,4],size=3,replace=False))
print(np.random.randint(1,10,size=(2)))
```
