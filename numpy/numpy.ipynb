导入numpy库

```python
import numpy as np
```

```python
a=[1,2,3,4,5,6,7,8]
a=np.reshape(a,(2,4))
print(a)
```

输出实例对象属性

```python
print(a.shape)
print(a.dtype)
print(a.ndim)
print(a.size)
```

矩阵运算

```python
x1=np.array([[1,2],[3,4]])
x2=np.array([[5,6],[7,8]])
sum=x1+x2
sub=x1-x2
mul=x1*x2
div=x1/x2
log=x1>x2
print(sum)
print(sub)
print(mul)
print(div)
print(log)
```

简单统计

```python
x3=np.array([11,22,33,4,55,66])
print(np.sum(x3))
print(np.mean(x3))
print(np.max(x3))
print(np.min(x3))
print(np.argmax(x3))
print(np.argmin(x3))
```
